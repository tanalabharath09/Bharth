```python
# ============================================================
# 🎓 STUDENT PERFORMANCE ANALYZER USING PCA
# Google Colab - Single Cell App
# ============================================================

!pip -q install gradio pandas numpy plotly

import numpy as np
import pandas as pd
import gradio as gr
import plotly.express as px
import plotly.graph_objects as go

# ============================================================
# DEFAULT STUDENT DATA
# ============================================================

subjects = [
    "Maths",
    "Physics",
    "Chemistry",
    "English",
    "Programming"
]

marks = [
    [82,78,85,74,88],
    [75,72,79,70,80],
    [91,88,94,82,96],
    [68,65,70,72,66],
    [85,82,88,78,91],
    [73,70,75,68,77],
    [95,92,96,85,98],
    [62,60,65,67,61],
    [79,76,82,73,84],
    [88,85,90,80,93],
    [71,68,74,69,73],
    [93,90,95,84,97],
    [66,64,69,71,65],
    [81,79,84,75,87],
    [76,74,78,72,79],
    [90,87,92,81,94],
    [64,62,67,69,63],
    [84,80,86,77,89],
    [72,69,76,70,75],
    [87,83,89,79,92],
    [69,66,72,68,70],
    [96,93,97,86,99],
    [78,75,81,74,82],
    [83,81,85,76,88],
    [67,63,68,70,64],
    [89,86,91,80,93],
    [74,71,77,69,78],
    [92,89,94,83,96],
    [80,77,83,75,85],
    [70,67,73,68,72]
]

default_df = pd.DataFrame(marks, columns=subjects)
default_df.insert(
    0,
    "Student",
    [f"Student {i}" for i in range(1,31)]
)

# ============================================================
# GLOBAL RESULT
# ============================================================

RESULT = {
    "scores": None,
    "students": []
}

# ============================================================
# PCA FUNCTION
# ============================================================

def calculate_pca(df):

    student_names = df.iloc[:,0].astype(str).tolist()

    X = df.iloc[:,1:].apply(
        pd.to_numeric,
        errors="coerce"
    )

    X = X.fillna(X.mean())

    X = X.to_numpy(dtype=float)

    # 1. Mean
    mean_values = np.mean(X, axis=0)

    # 2. Centering
    Xc = X - mean_values

    # 3. Covariance matrix
    covariance = np.cov(
        Xc,
        rowvar=False
    )

    # 4. Eigenvalues and eigenvectors
    eigenvalues, eigenvectors = np.linalg.eigh(
        covariance
    )

    # Sort from largest to smallest
    order = np.argsort(eigenvalues)[::-1]

    eigenvalues = eigenvalues[order]
    eigenvectors = eigenvectors[:,order]

    # 5. Explained variance
    total = np.sum(eigenvalues)

    if total != 0:
        explained = (
            eigenvalues / total
        ) * 100
    else:
        explained = np.zeros_like(eigenvalues)

    # 6. First two principal components
    W = eigenvectors[:,:2]

    # 7. Project students
    scores = Xc @ W

    # DataFrames
    centered_df = pd.DataFrame(
        Xc,
        columns=df.columns[1:]
    )

    covariance_df = pd.DataFrame(
        covariance,
        index=df.columns[1:],
        columns=df.columns[1:]
    )

    eigenvalue_df = pd.DataFrame({
        "Component":
            [f"PC{i+1}" for i in range(len(eigenvalues))],

        "Eigenvalue":
            eigenvalues,

        "Explained Variance (%)":
            explained,

        "Cumulative Variance (%)":
            np.cumsum(explained)
    })

    eigenvector_df = pd.DataFrame(
        eigenvectors,
        index=df.columns[1:],
        columns=[
            f"PC{i+1}"
            for i in range(len(eigenvalues))
        ]
    )

    score_df = pd.DataFrame({
        "Student": student_names,
        "PC1": scores[:,0],
        "PC2": scores[:,1]
    })

    return (
        centered_df,
        covariance_df,
        eigenvalue_df,
        eigenvector_df,
        score_df,
        explained
    )

# ============================================================
# RUN ANALYSIS
# ============================================================

def run_analysis(data):

    global RESULT

    try:

        if data is None:
            return (
                "Please enter student data.",
                None,
                None,
                None,
                None,
                None,
                None,
                None,
                None,
                None
            )

        df = pd.DataFrame(data)

        # Remove completely empty rows
        df = df.dropna(how="all")

        if len(df.columns) < 3:
            return (
                "Please provide Student column and at least 2 subjects.",
                None,None,None,None,None,None,None,None,None
            )

        # Student names
        df.iloc[:,0] = df.iloc[:,0].fillna("")

        for i in range(len(df)):
            if str(df.iloc[i,0]).strip() == "":
                df.iloc[i,0] = f"Student {i+1}"

        # Convert marks to numbers
        for col in df.columns[1:]:
            df[col] = pd.to_numeric(
                df[col],
                errors="coerce"
            )

        # Remove empty subject columns
        valid_columns = []

        for col in df.columns[1:]:
            if df[col].notna().sum() > 0:
                valid_columns.append(col)

        df = df[
            [df.columns[0]] + valid_columns
        ]

        if len(valid_columns) < 2:
            return (
                "At least 2 subjects are required.",
                None,None,None,None,None,None,None,None,None
            )

        # Missing values
        for col in valid_columns:

            mean_value = df[col].mean()

            if pd.isna(mean_value):
                mean_value = 0

            df[col] = df[col].fillna(
                mean_value
            )

            # Marks 0-100
            df[col] = df[col].clip(
                0,
                100
            )

        # PCA
        (
            centered,
            covariance,
            eigenvalue,
            eigenvector,
            scores,
            explained
        ) = calculate_pca(df)

        RESULT["scores"] = scores
        RESULT["students"] = scores["Student"].tolist()

        # ====================================================
        # SUMMARY
        # ====================================================

        pc1 = explained[0]
        pc2 = explained[1]
        top2 = pc1 + pc2

        summary = (
            "# Student Performance Analyzer\n\n"
            "## Analysis Completed Successfully\n\n"
            f"**Number of Students:** {len(df)}\n\n"
            f"**Number of Subjects:** {len(valid_columns)}\n\n"
            f"**PC1 Explained Variance:** {pc1:.2f}%\n\n"
            f"**PC2 Explained Variance:** {pc2:.2f}%\n\n"
            f"**Top 2 PCs:** {top2:.2f}%\n\n"
            "### PCA Process\n\n"
            "1. Student data collected\n"
            "2. Mean calculated\n"
            "3. Data centered\n"
            "4. Covariance matrix calculated\n"
            "5. Eigenvalues calculated\n"
            "6. Eigenvectors calculated\n"
            "7. Top 2 Principal Components selected\n"
            "8. Students projected into 2D\n\n"
            "### Meaning\n\n"
            "Students close together in the PCA graph "
            "have similar performance patterns.\n\n"
            "Students far apart have different "
            "performance patterns."
        )

        # ====================================================
        # PCA SCATTER PLOT
        # ====================================================

        fig_pca = px.scatter(
            scores,
            x="PC1",
            y="PC2",
            text="Student",
            hover_name="Student",
            title="Students in PCA 2D Space"
        )

        fig_pca.update_traces(
            textposition="top center"
        )

        fig_pca.update_layout(
            height=600
        )

        # ====================================================
        # VARIANCE GRAPH
        # ====================================================

        variance_df = eigenvalue.copy()

        fig_variance = px.bar(
            variance_df,
            x="Component",
            y="Explained Variance (%)",
            text="Explained Variance (%)",
            title="Explained Variance"
        )

        fig_variance.update_traces(
            texttemplate="%{text:.2f}%",
            textposition="outside"
        )

        # ====================================================
        # SUBJECT CONTRIBUTION GRAPH
        # ====================================================

        loadings = pd.DataFrame({
            "Subject": df.columns[1:],
            "PC1": eigenvector["PC1"],
            "PC2": eigenvector["PC2"]
        })

        fig_loading = go.Figure()

        fig_loading.add_trace(
            go.Bar(
                x=loadings["Subject"],
                y=loadings["PC1"],
                name="PC1"
            )
        )

        fig_loading.add_trace(
            go.Bar(
                x=loadings["Subject"],
                y=loadings["PC2"],
                name="PC2"
            )
        )

        fig_loading.update_layout(
            title="Subject Contribution",
            barmode="group",
            height=500
        )

        return (
            summary,
            df,
            centered,
            covariance,
            eigenvalue,
            eigenvector,
            scores,
            fig_pca,
            fig_variance,
            fig_loading
        )

    except Exception as e:

        error = (
            "Error occurred: "
            + str(e)
        )

        return (
            error,
            None,None,None,None,
            None,None,None,None,None
        )

# ============================================================
# FIND SIMILAR STUDENTS
# ============================================================

def find_similar(student):

    if RESULT["scores"] is None:
        return "Please run PCA analysis first."

    scores = RESULT["scores"]

    if student not in scores["Student"].values:
        return "Student not found."

    selected = scores[
        scores["Student"] == student
    ].iloc[0]

    selected_point = np.array([
        selected["PC1"],
        selected["PC2"]
    ])

    distances = []

    for index,row in scores.iterrows():

        if row["Student"] == student:
            continue

        point = np.array([
            row["PC1"],
            row["PC2"]
        ])

        distance = np.linalg.norm(
            selected_point - point
        )

        distances.append({
            "Student": row["Student"],
            "Distance": distance
        })

    distance_df = pd.DataFrame(
        distances
    )

    distance_df = distance_df.sort_values(
        "Distance"
    ).head(5)

    text = (
        f"# Similar Students to {student}\n\n"
    )

    for i,row in distance_df.iterrows():

        text += (
            f"**{row['Student']}**  "
            f"Distance = {row['Distance']:.3f}\n\n"
        )

    return text

# ============================================================
# DOWNLOAD RESULTS
# ============================================================

def download_results():

    if RESULT["scores"] is None:
        return None

    path = "/content/PCA_Student_Results.csv"

    RESULT["scores"].to_csv(
        path,
        index=False
    )

    return path

# ============================================================
# RESET
# ============================================================

def reset_data():

    return default_df

# ============================================================
# APP CSS
# ============================================================

css = (
    ".gradio-container {"
    "max-width: 1400px !important;"
    "}"
)

# ============================================================
# CREATE APP
# ============================================================

with gr.Blocks(
    title="Student Performance Analyzer",
    css=css,
    theme=gr.themes.Soft()
) as app:

    gr.Markdown(
        "# 🎓 Student Performance Analyzer\n"
        "### Principal Component Analysis (PCA) Based App\n\n"
        "Analyze student performance using mathematics, "
        "PCA and interactive visualizations."
    )

    # ========================================================
    # DATA TAB
    # ========================================================

    with gr.Tab("📋 Student Data"):

        gr.Markdown(
            "## Edit Student Marks\n\n"
            "Change the marks below and click "
            "**RUN PCA ANALYSIS**."
        )

        data_table = gr.Dataframe(
            value=default_df,
            headers=list(default_df.columns),
            datatype=["str"] + ["number"] * 5,
            row_count=(30,"dynamic"),
            col_count=(6,"fixed"),
            interactive=True,
            label="Student Marks"
        )

        with gr.Row():

            run_button = gr.Button(
                "🚀 RUN PCA ANALYSIS",
                variant="primary"
            )

            reset_button = gr.Button(
                "🔄 RESET DATA"
            )

    # ========================================================
    # DASHBOARD
    # ========================================================

    with gr.Tab("📊 Dashboard"):

        summary = gr.Markdown(
            "Run PCA analysis to see results."
        )

        with gr.Row():

            pca_graph = gr.Plot(
                label="PCA Map"
            )

            variance_graph = gr.Plot(
                label="Explained Variance"
            )

        loading_graph = gr.Plot(
            label="Subject Contribution"
        )

    # ========================================================
    # MATHEMATICS
    # ========================================================

    with gr.Tab("🧮 PCA Mathematics"):

        gr.Markdown(
            "## Original Data"
        )

        original_table = gr.Dataframe(
            interactive=False
        )

        gr.Markdown(
            "## Centered Data"
        )

        centered_table = gr.Dataframe(
            interactive=False
        )

        gr.Markdown(
            "## Covariance Matrix"
        )

        covariance_table = gr.Dataframe(
            interactive=False
        )

        gr.Markdown(
            "## Eigenvalues and Explained Variance"
        )

        eigenvalue_table = gr.Dataframe(
            interactive=False
        )

        gr.Markdown(
            "## Eigenvectors"
        )

        eigenvector_table = gr.Dataframe(
            interactive=False
        )

        gr.Markdown(
            "## Student PCA Coordinates"
        )

        score_table = gr.Dataframe(
            interactive=False
        )

    # ========================================================
    # SIMILAR STUDENTS
    # ========================================================

    with gr.Tab("👥 Similar Students"):

        gr.Markdown(
            "## Find Similar Students\n\n"
            "Choose a student and find the 5 students "
            "closest to them in PCA space."
        )

        student_dropdown = gr.Dropdown(
            choices=[
                f"Student {i}"
                for i in range(1,31)
            ],
            value="Student 1",
            label="Select Student"
        )

        similar_button = gr.Button(
            "🔍 FIND SIMILAR STUDENTS",
            variant="primary"
        )

        similar_result = gr.Markdown()

    # ========================================================
    # EXPORT
    # ========================================================

    with gr.Tab("💾 Export"):

        gr.Markdown(
            "## Download PCA Results\n\n"
            "Run the analysis first, then download "
            "the PCA coordinates."
        )

        download_button = gr.Button(
            "⬇️ DOWNLOAD RESULTS",
            variant="primary"
        )

        download_file = gr.File(
            label="PCA Results"
        )

    # ========================================================
    # ABOUT
    # ========================================================

    with gr.Tab("ℹ️ About Project"):

        gr.Markdown(
            "# About the Project\n\n"
            "## What is PCA?\n\n"
            "Principal Component Analysis is a mathematical "
            "method used to reduce dimensions while preserving "
            "important information.\n\n"
            "### Mathematical Steps\n\n"
            "**1. Data Matrix**\n\n"
            "Student × Subject\n\n"
            "**2. Centering**\n\n"
            "Xc = X - Mean\n\n"
            "**3. Covariance Matrix**\n\n"
            "C = (1 / (n - 1)) Xcᵀ Xc\n\n"
            "**4. Eigenvalue Equation**\n\n"
            "Cv = λv\n\n"
            "**5. Principal Components**\n\n"
            "The eigenvectors corresponding to the largest "
            "eigenvalues are selected as principal components.\n\n"
            "### Real World Use\n\n"
            "• Find similar students\n\n"
            "• Understand performance patterns\n\n"
            "• Visualize high-dimensional data\n\n"
            "• Identify important subjects\n\n"
            "• Support student performance analysis"
        )

    # ========================================================
    # BUTTON ACTIONS
    # ========================================================

    run_button.click(
        fn=run_analysis,
        inputs=data_table,
        outputs=[
            summary,
            original_table,
            centered_table,
            covariance_table,
            eigenvalue_table,
            eigenvector_table,
            score_table,
            pca_graph,
            variance_graph,
            loading_graph
        ]
    )

    reset_button.click(
        fn=reset_data,
        inputs=None,
        outputs=data_table
    )

    similar_button.click(
        fn=find_similar,
        inputs=student_dropdown,
        outputs=similar_result
    )

    download_button.click(
        fn=download_results,
        inputs=None,
        outputs=download_file
    )

# ============================================================
# START APP
# ============================================================

app.launch(
    share=True
)
```
